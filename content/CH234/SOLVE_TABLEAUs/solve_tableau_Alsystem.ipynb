```octave
AlT=1e-1;

TABLEAU=[...
%H+   Al     logK
1     0      0       %H
0     1      0       %Al
-1    0      -14     %OH
-1    1      -4.99   %AlOH
-2    1      -9.3    %AlOH2
-3    1      -15     %AlOH3
-4    1      -23     %AlOH4
-32   13     -111.5  %Al13OH32
0     AlT   0       %TOTALS
];

[N,M]=size(TABLEAU);T=TABLEAU(N,1:M-1)'; T(T==0)=1e-30; Xguess=T./10;

[Xans,maxerror]=get_solution(TABLEAU,Xguess)
pH=-log10(Xans(1))

if maxerror > max(T)*1e-4;
    [n,m]=size(TABLEAU);
    subTABLEAU=TABLEAU(1:n-2,1:m); % remove last species 
    subTABLEAU(n-1,1:m)=TABLEAU(n,1:m);% put total vector back
    [Xans,maxerror]=get_solution(subTABLEAU,Xguess)
    Xguess=Xans;
    pH=-log10(Xans(1))
end

[Xans,maxerror]=get_solution(TABLEAU,Xguess)
pH=-log10(Xans(1))
```
