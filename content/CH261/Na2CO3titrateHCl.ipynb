# Titration of 100 mL of 0.01 M NaCO with 0.1 M HCl.
For this we just take a fixed pH and solve for each of the charged species.  Rearrange electroneutrality to solve for [Cl-] and convert that to volume.

```octave
NaT=0.02; pH=11.1:-0.1:2; Kw=1e-14; CT=0.01; Ka1=10^-6.3; Ka2=10^-10.3; VT=0.1; HClconc=0.1;

for i=1:length(pH)
    H=10^(-1*pH(i));
    OH=Kw/H;
    HCO3=(CT*Ka1*H)/(H^2+Ka1*H+Ka1*Ka2);
    CO3=(CT*Ka1*Ka2)/(H^2+Ka1*H+Ka1*Ka2);
    Cl=NaT+H-OH-HCO3-2*CO3;
    vol(i)=1000*((Cl*VT)/HClconc);
end

plot(vol,pH,'linewidth',2)
set(gca,'fontsize',11,'linewidth',2)
xlabel('volume (mL)'); ylabel('pH');
axis([0 30 2 12])

```
