# Strong-Strong Titrations

If you start with 0.01 M NaOH the initial pH will be 12 (-log(1e-14/1e-12)=2).  As you add strong acid the pH will decrease.  Neutralizing the base in this way is called titration.  At any point during the titration the charge balance must be satisfied ...

$[\mathrm{H}^+]+[\mathrm{Na}^+]=[\mathrm{OH}^-]+[\mathrm{Cl}^-]$

Using K$_w$ we can write this in terms of H+ and $\mathrm{Na_T}$

$[\mathrm{H}^+]+\mathrm{Na_T}=\dfrac{K_w}{[\mathrm{H}^+]}+[\mathrm{Cl}^-]$

We can assume that the concentration of the added acid is high enough that we can ignore dilution (small volumes are being added) so $[\mathrm{Na}^+]$ = $\mathrm{Na_T}$= 0.01 M, for the entire titration (if we knew the volumes we could calculate the exact concentration of this strong acid anion, but I am trying to keep the calculations simple and illustrate the principles of titration and buffering, and electroneutrality).  

So with total Na+ known we have just two variables, the chloride concentration and the proton concentration.  If we set one variable we have one equation and one unknown and we can solve it.  To make a graph of the pH versus added chloride it is actually easier to fix pH ($[\mathrm{H}^+]$) and solve for $[\mathrm{Cl}^-]$. So by rearranging the electroneutrality we can write 

$[\mathrm{Cl}^-]=[\mathrm{H}^+]+\mathrm{Na_T}-\dfrac{K_w}{[\mathrm{H}^+]}$

In the code below we'll do this calculation and plot the titration curve

```octave
%plot -s 600,600 -f 'svg'
Kw=1e-14; pH=12:-0.1:2; H=10.^-pH; OH=Kw./H; NaT=0.01;
Cl=NaT+H-OH;
plot(Cl,pH,'linewidth',2); xlabel('[Cl^-]'); ylabel('pH')
set(gca,'fontsize',11,'linewidth',2)
title('NaOH titrated with HCl')
```

# convert to volume

of course we don't titrate concentration directly.  we use volumes of titrant added.  The volume we can get from 

$ C= \dfrac{n}{V} $

rearrange

$ V=\dfrac{n}{C} $

here C is 0.1 mol/L (titrant) and n is the moles at the point in the titration.  we get that from the concentration we solved for rearranging the same expression (But for the beaker this time) n=[Cl-]*(0.100). For the 100 mL beaker.


```octave
%plot -s 600,600 -f 'svg'
Kw=1e-14; pH=12:-0.1:2; H=10.^-pH; OH=Kw./H; NaT=0.01;
volCl=(Cl*0.1)/(0.1); volCl=volCl*1000; % convert to mL
plot(volCl,pH,'linewidth',2); xlabel('Volume (mL)'); ylabel('pH')
set(gca,'fontsize',11,'linewidth',2)
title('NaOH titrated with HCl')

specialV=[0 10 20];
specialpH=[12 7 2];
hold on
plot(specialV,specialpH,'ko','markersize',12)
```
